```python
db = DataBaseOperation()
studentvle = db.fetch_all_as_dataframe("""
SELECT TOP (100) 
	[code_module]
    ,[code_presentation]
    ,[id_student]
    ,[id_site]
    ,[date]
    ,[sum_click]
FROM [OULAD].[dbo].[studentVle]
""")

courses = db.fetch_all_as_dataframe("""
SELECT TOP (100)
	[code_module]
    ,[code_presentation]
    ,[module_presentation_length]
FROM [OULAD].[dbo].[courses]
""")

assessments = db.fetch_all_as_dataframe("""
SELECT TOP (100)
	[code_module]
    ,[code_presentation]
    ,[id_assessment]
    ,[assessment_type]
    ,[date]
    ,[weight]
FROM [OULAD].[dbo].[assessments]
""")

vle = db.fetch_all_as_dataframe("""
SELECT TOP (100)
	[id_site]
    ,[code_module]
    ,[code_presentation]
    ,[activity_type]
    ,[week_from]
    ,[week_to]
FROM [OULAD].[dbo].[vle]
""")

studentInfo = db.fetch_all_as_dataframe("""
SELECT TOP (100)
	[code_module]
    ,[code_presentation]
    ,[id_student]
    ,[gender]
    ,[region]
    ,[highest_education]
    ,[imd_band]
    ,[age_band]
    ,[num_of_prev_attempts]
    ,[studied_credits]
    ,[disability]
    ,[final_result]
FROM [OULAD].[dbo].[studentInfo]
""")

studentRegistration = db.fetch_all_as_dataframe("""
SELECT TOP (100) 
	[code_module]
    ,[code_presentation]
    ,[id_student]
    ,[date_registration]
    ,[date_unregistration]
FROM [OULAD].[dbo].[studentRegistration]
""")

studentAssessment = db.fetch_all_as_dataframe("""
SELECT TOP (100) 
	[id_assessment]
    ,[id_student]
    ,[date_submitted]
    ,[is_banked]
    ,[score]
FROM [OULAD].[dbo].[studentAssessment]
""")

studentVle = db.fetch_all_as_dataframe("""
SELECT TOP (100) 
	[code_module]
    ,[code_presentation]
    ,[id_student]
    ,[id_site]
    ,[date]
    ,[sum_click]
FROM [OULAD].[dbo].[studentVle]
""")
```
